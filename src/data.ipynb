```python
from torchvision.transforms.functional import to_pil_image
import torch
import random
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
```

```python
def visualize(image, bboxes, masks, figsize=(10, 10)):
    if isinstance(image, torch.Tensor):
        image = to_pil_image(image)
    if isinstance(bboxes, torch.Tensor):
        bboxes = bboxes.numpy()
    if isinstance(masks, torch.Tensor):
        masks = masks.numpy()

    if len(bboxes.shape) == 1:  # Nếu chỉ có một bounding box
        bboxes = [bboxes]

    print("Bboxes:", bboxes)  # In ra bboxes để debug

    fig, ax = plt.subplots(1, figsize=figsize)
    ax.imshow(image)

    for bbox in bboxes:
        if len(bbox) == 4:  # Kiểm tra bounding box có 4 phần tử
            x1, y1, x2, y2 = bbox
            color = [random.random() for _ in range(3)]
            rect = patches.Rectangle((x1, y1), x2 - x1, y2 - y1, linewidth=2, edgecolor=color, facecolor='none')
            ax.add_patch(rect)

    for mask in masks:
        color = np.array([random.random() for _ in range(3)] + [0.5])
        masked_image = np.zeros_like(image, dtype=np.float32)
        for c in range(3):
            masked_image[..., c] = mask * color[c]
        ax.imshow(masked_image, alpha=0.5)

    plt.axis('off')
    plt.show()
```
